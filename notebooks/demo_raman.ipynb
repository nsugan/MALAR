# MALAR — Raman demo
Train a Raman virus DomainSpec to coverage, then infer on a spectrum.

```python
from malar.training.domainspec import load_domainspec
from malar.training.campaign import Campaign
spec=load_domainspec('domains/raman_virus.yaml')
camp=Campaign(spec)
report=camp.run(verbose=True)
print('stop:',report.stop_reason,'met:',report.targets_met)
```

```python
from malar.inference.service import InferenceService
import numpy as np
from malar.world.adapters.raman import _CLASS_PEAKS,_spectrum
svc=InferenceService(camp.engine)
rng=np.random.default_rng(0)
img=np.array([_spectrum(_CLASS_PEAKS['sars_cov_2'],128,rng) for _ in range(20)])
print(svc.infer(image=img))
print(svc.infer(text='unknown gibberish'))
```
