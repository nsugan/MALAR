# MALAR — synthetic demo
Run the deterministic loop over synthetic blobs/circles and watch memory grow then stabilise.

```python
from malar.world.adapters.synthetic import SyntheticAdapter
from malar.core.loop import MalarEngine
ad=SyntheticAdapter(n_ticks=14,points_per_tick=40,k=6)
eng=MalarEngine(ad,world_id='synthetic',budget=32)
sizes=[]
for b in ad.stream():
    st=eng.step(b)
    sizes.append(eng.memory_size())
    print(st.t, st.objects[0]['class'], st.curator_decision.action, sizes[-1])
print('growth:',sizes)
```
